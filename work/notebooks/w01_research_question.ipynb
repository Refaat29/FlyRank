## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*
Answer:


**What my work CAN say:**
- Observed patterns: for example, declining pages in this dataset show a
  lower average CTR than stable or growing pages.
- Directional signals: certain measurable features (impressions, CTR,
  content age) appear associated with decline, based on the data alone.
- Decision-support: a ranked list of pages that a human reviewer should
  check first, based on evidence — not a guarantee that fixing them will
  work.

**What my work CANNOT say:**
- I cannot claim my model or analysis "proves" why a page is declining,
  since this is observational data, not an experiment.
- I cannot claim that refreshing a page will cause it to recover — that
  would require a controlled test, which this data does not provide.
- I cannot claim to have discovered or predicted anything about Google's
  actual ranking algorithm — I only observe outcomes like impressions,
  clicks, and CTR, not the mechanism behind them.
- I cannot treat correlation (e.g. lower CTR on declining pages) as proof
  of causation.
